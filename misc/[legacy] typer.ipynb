```python
import pyperclip

code_map = {
    "a":  "\ue000", "a2": "\ue001", "a3": "\ue002",
    "e":  "\ue003", "e2": "\ue004",
    "i":  "\ue005",
    "o":  "\ue006", "o2": "\ue007",
    "u":  "\ue008", "u2": "\ue009",
    "ie2":"\ue00a", "u2o3":"\ue00b", "uo2":"\ue00c",

    "ik": "\ue00e", "ok": "\ue00f",

    "b":  "\ue011", "c":  "\ue012", "d":  "\ue013", "d2": "\ue014",
    "g":  "\ue015", "h":  "\ue016", "l":  "\ue017", "m":  "\ue018",
    "n":  "\ue019", "t":  "\ue01a", "v":  "\ue01b", "x":  "\ue01c",
    "ch": "\ue01d", "kh": "\ue01e", "ng": "\ue01f", "nh": "\ue020",
    "ph": "\ue021", "th": "\ue022",

    "mk": "\ue024", "tk": "\ue025",
}

def encode_custom(text):
    tokens = text.split()
    return "".join(code_map.get(tok, tok) for tok in tokens)
```

```python
print("Converter running.")

exit_program = False
while not exit_program:
    user_input = input("Enter codes: ").strip()
    if not user_input:
        exit_program = True

    encoded = encode_custom(user_input)
    pyperclip.copy(encoded)
    print("Copied to clipboard.")
```

```python
import tkinter as tk

code_map = {
    "a":  "\ue000", "a2": "\ue001", "a3": "\ue002",
    "e":  "\ue003", "e2": "\ue004",
    "i":  "\ue005",
    "o":  "\ue006", "o2": "\ue007",
    "u":  "\ue008", "u2": "\ue009",
    "ie2":"\ue00a", "u2o3":"\ue00b", "uo2":"\ue00c",

    "ik": "\ue00e", "ok": "\ue00f",

    "b":  "\ue011", "c":  "\ue012", "d":  "\ue013", "d2": "\ue014",
    "g":  "\ue015", "h":  "\ue016", "l":  "\ue017", "m":  "\ue018",
    "n":  "\ue019", "t":  "\ue01a", "v":  "\ue01b", "x":  "\ue01c",
    "ch": "\ue01d", "kh": "\ue01e", "ng": "\ue01f", "nh": "\ue020",
    "ph": "\ue021", "th": "\ue022",

    "mk": "\ue024", "tk": "\ue025",
}

def encode_custom(text):
    return "".join(code_map.get(tok, tok) for tok in text.split())

def on_text_change(event=None):
    encoded = encode_custom(entry.get())
    output_label.config(text=encoded)

root = tk.Tk()
root.title("PUA Font Encoder")

entry = tk.Entry(root, font=("Segoe UI", 14), width=40)
entry.pack(padx=20, pady=10)
entry.bind("<KeyRelease>", on_text_change)

output_label = tk.Label(
    root,
    text="",
    font=("Khoa Dau", 48),
    bg="white"
)
output_label.pack(padx=20, pady=20, fill="both")

on_text_change()
root.mainloop()
```
